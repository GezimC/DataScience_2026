### Bonus 1

Find the top 10 jobs by average purchase price, but only include jobs that appear at least 5 times.

### Bonus 2

Create a new column:

`purchase_category`

Rules:

- Low: purchase price < 25
- Medium: 25 to 75
- High: above 75

Then count how many purchases are in each category.

### Bonus 3

Find the most common email domain for customers with high purchases.

```python
# Bonus 1 solution idea

job_summary = (
    df_clean.groupby("job")
    .agg(
        purchase_count=("purchase_price", "count"),
        avg_purchase_price=("purchase_price", "mean")
    )
    .reset_index()
)

top_jobs_avg_price = (
    job_summary[job_summary["purchase_count"] >= 5]
    .sort_values("avg_purchase_price", ascending=False)
    .head(10)
)

top_jobs_avg_price
```

```python
# Bonus 2 solution idea

df_clean["purchase_category"] = pd.cut(
    df_clean["purchase_price"],
    bins=[-1, 25, 75, df_clean["purchase_price"].max()],
    labels=["Low", "Medium", "High"]
)

df_clean["purchase_category"].value_counts()
```

```python
# Bonus 3 solution idea

high_purchase_domains = (
    df_clean[df_clean["purchase_category"] == "High"]["email_domain"]
    .value_counts()
    .head(10)
)

high_purchase_domains
```
