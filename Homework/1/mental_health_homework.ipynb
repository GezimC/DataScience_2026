## 🔹 Setup

```
import pandas as pd
import seaborn as sns
import plotly.express as px

sns.set_theme(style="whitegrid")

df = pd.read_csv("../../Projects/teen_mental_health_dataset.csv")
df.head()
```

### Q1: Basic Overview

```
print(df.shape)
print(df.columns)
print(df.dtypes)
```

### Q2: Missing Values

```
df.isnull().sum().sort_values(ascending=False)
```

### Q3: Identify Variables
- Which are numerical?
- Which are categorical?
- What could be target variable (stress/anxiety/depression)?

```
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include=["object", "category", "string"]).columns

print("Numerical:", num_cols.tolist())
print("Categorical:", cat_cols.tolist())
```

# 🔹 PART 2 — Data Cleaning

### Q4: Handle Missing Values
Decide: drop or fill. Explain why.

```
df_clean = df.dropna()
df_clean.shape
```

### Q5: Check Duplicates

```
df_clean.duplicated().sum()
```

# 🔹 PART 3 — Feature Engineering

### Q6: Create Usage Category
- Low (<2h)
- Medium (2–5h)
- High (>5h)

```
def create_usage_categories(hours):
    if hours <2:
        return 'Low'
    elif hours >2 and hours <5:
        return 'Medium'
    else:
        return 'High'

df['usage_cat']= df_clean['daily_social_media_hours'].apply(create_usage_categories)
```

### Q9: Group Analysis — Sleep vs Anxiety

# 🔹 PART 6 — Seaborn Visualization

### Q13: Distribution of Social Media Hours

```
sns.histplot(df_clean["Social_Media_Hours"], kde=True)
```

### Q14: Stress by Usage Category

```
sns.boxplot(data=df_clean, x="usage_category", y="Stress_Level")
```

# 🔹 PART 7 — Plotly (ONLY 2 QUESTIONS)

### Q17: Interactive Scatter

```
fig = px.scatter(
    df_clean,
    x="Social_Media_Hours",
    y="Stress_Level",
    color="usage_category",
    title="Social Media vs Stress"
)
fig.show()
```

### Q18: Interactive Bar

```
avg_stress = df_clean.groupby("usage_category")["Stress_Level"].mean().reset_index()

fig = px.bar(
    avg_stress,
    x="usage_category",
    y="Stress_Level",
    title="Average Stress by Usage"
)
fig.show()
```
